```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Cleaned_Dataset.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

values = df['Price range'].value_counts().sort_index(ascending=True)
price_index = values.index

plt.bar(price_index, values)
plt.xticks(price_index)
plt.xlabel('Price Range')
plt.ylabel('Value Count')
plt.show()
```

## Insight
Order of Price range counts across the dataset is 1 > 2 > 3 > 4. Therefore, 1,2 price ranges are the more common.

```python
price_group = df.groupby('Price range').agg({'Aggregate rating': ['mean', 'median']})
price_group
```

## Insight

1. Restaurants with price range of 1 on an avg have approx 2 rating.
2. Restaurants with price range of 2 on an avg have approx 2.9 rating.
3. Restaurants with price range of 3 on an avg have approx 3.7 rating.
4. Restaurants with price range of 4 on an avg have approx 3.8 rating.
