```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Cleaned_Dataset.csv')
df.head()
```

```python
df['Restaurant Name Length'] = df['Restaurant Name'].str.len()
df.head()
```

```python
df['Restaurant Name'].duplicated().sum()
# couldn't generalize the cuisines since duplicates are present
```

```python
df['Amount of Cuisines Offered'] = df['Cuisines'].str.split(',')
df['Amount of Cuisines Offered'] = df['Amount of Cuisines Offered'].str.len()
df.head()
```

```python
df['Offers Table Booking'] = df['Has Table booking'].map({
    1: 'Yes',
    0: 'No'})
df.head()
```

```python
df['Offers Online delivery'] = df['Has Online delivery'].map({
    1: 'Yes',
    0: 'No'})
df.head()
```

```python
df.drop(columns=['Has Online delivery', 'Has Table booking'], inplace=True)
df.head()
```

```python
df['Currency'].value_counts().index
```

```python
currency_symbols = {
    'Indian Rupees(Rs.)': '₹',
    'Dollar($)': '$',
    'Pounds(£)': '£',
    'Brazilian Real(R$)': 'R$',
    'Emirati Diram(AED)': 'د.إ',
    'Rand(R)': 'R',
    'NewZealand($)': 'NZ$',
    'Turkish Lira(TL)': '₺',
    'Botswana Pula(P)': 'P',
    'Indonesian Rupiah(IDR)': 'Rp',
    'Qatari Rial(QR)': 'ر.ق',
    'Sri Lankan Rupee(LKR)': 'Rs.'
}
df['Currency Symbol'] = df['Currency'].map(currency_symbols)
df
```

```python
average_cost = df['Average Cost for two']
average_cost.head()
```

```python
average_cost.min()
```

```python
average_cost.max()
```

```python
exchange_rates = {
    'Indian Rupees(Rs.)': 1,
    'Dollar($)': 83,
    'Pounds(��)': 105,
    'Brazilian Real(R$)': 16,
    'Emirati Diram(AED)': 22.5,
    'Rand(R)': 4.5,
    'NewZealand($)': 50,
    'Turkish Lira(TL)': 2.5,
    'Botswana Pula(P)': 6,
    'Indonesian Rupiah(IDR)': 0.005,
    'Qatari Rial(QR)': 22.8,
    'Sri Lankan Rupee(LKR)': 0.27
}
# converting to INR.
df['exchange value'] = df['Currency'].map(exchange_rates)
df['INR value'] = df['exchange value']*df['Average Cost for two']
df.head()
```

```python
bins = [0, 500, 1500, 3000, 7000, 20000, float('inf')] # binning
labels = ['Inexpensive','Budget','Moderate','Expensive','Premium','Luxury']
df['Price Category'] = pd.cut(df['INR value'],bins=bins,labels=labels,include_lowest=True)
df.head()
```

```python
df['Is delivering now'] = df['Is delivering now'].map({
    0:'No',
    1:'Yes'
})
df['Address length'] = df['Address'].str.len()
df.head()
```

```python
df.to_csv('Featured_data.csv')
```
