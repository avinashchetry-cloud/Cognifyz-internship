```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Cleaned_Dataset.csv')
df.head()
```

```python
booking_df = df[df['Has Table booking'] == 1][['Restaurant Name', 'Restaurant ID', 'Has Online delivery', 'Has Table booking']]

booking_df.head()
```

```python

print(f'Percentage of restaurants who provide booking delivery is: {round(((booking_df.shape[0]/df.shape[0])*100), 2)}%')
```

```python
online_df = df[df['Has Online delivery'] == 1][['Restaurant Name', 'Restaurant ID', 'Has Online delivery', 'Has Table booking']]
online_df.head()
```

```python
print(f'Percentage of restaurants who provide online delivery is: {round(((online_df.shape[0]/df.shape[0])*100), 2)}%')
```

```python
both_services_df = df[(df['Has Online delivery'] == 1) & (df['Has Table booking'] == 1)][['Restaurant Name', 'Restaurant ID', 'Has Online delivery', 'Has Table booking', 'Price range']]
both_services_df.head()
```

```python
print(f'Percentage of restaurants who provide both services is: {round((both_services_df.shape[0]/df.shape[0])*100, 2)}%')
```

```python
table_booking = df.groupby('Has Table booking').agg({'Aggregate rating': ['mean', 'median']})
ranges_df = pd.DataFrame(columns=['1', '2', '3', '4'])

def price_range(row):
    ranges_df.loc[len(ranges_df)] = [None, None, None, None]
    ranges_df.loc[ranges_df.index[-1], str(row['Price range'])] = row['Restaurant Name']

both_services_df.apply(price_range, axis=1)
ranges_df
```
