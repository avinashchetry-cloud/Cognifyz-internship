```python
numeric_cols = df.drop('Restaurant ID', axis=1).select_dtypes(include=['number'])
cate_cols = df.drop(columns=['Restaurant Name', 'Address', 'Locality', 'Locality Verbose', 'Cuisines', 'City', 'Currency', 'Switch to order menu']).select_dtypes(include=['str'])
for col in numeric_cols.columns:
    print(col + ':\n')
    print("Mean: ", df[col].mean())
    print("Median: ", df[col].median())
    print("Standard Deviation: ", df[col].std())
    print("Kurtosis: ", df[col].kurtosis())
    print("Skew: ", df[col].skew() , '\n')
```

```python
import matplotlib.pyplot as plt

for col in cate_cols:
    print(col, '\n')
    print(df[col].value_counts().sort_values(ascending=True), '\n')
```

```python
import matplotlib.pyplot as plt

for col in cate_cols:
    x_axis = df[col].value_counts().sort_values(ascending=False).index
    y_axis = df[col].value_counts().sort_values(ascending=False)

    plt.bar(x_axis, y_axis)
    plt.title(f" {col} distribution")
    plt.show()
```

```python
print("Restaurant Name:", df['Restaurant Name'].mode()[0])
print("Country Code:", df['Country Code'].mode()[0])
print("City: ", df['City'].mode()[0])
print("Cuisines: ", df['Cuisines'].mode()[0])
print("Currency: ", df['Currency'].mode()[0])
```

```python
cuisines = df.assign(Cuisines=df['Cuisines'].str.split(','))
cuisines = cuisines.explode('Cuisines')
cuisines['Cuisines'] = cuisines['Cuisines'].str.strip()
cuisines_group = cuisines.groupby('Cuisines').agg({'Restaurant Name': 'nunique'}).sort_values('Restaurant Name', ascending=False)
cuisines_group.rename(columns={'Restaurant Name': 'Restaurant count'}, inplace=True)
cuisines_x = cuisines_group.sort_values('Restaurant count', ascending=False)['Restaurant count'].head(10).index
cuisines_y = cuisines_group.sort_values('Restaurant count', ascending=False)['Restaurant count'].head(10)

plt.bar(cuisines_x, cuisines_y)
plt.title('Top 10 Cuisines')
plt.xticks(rotation=90)
plt.show()
```
