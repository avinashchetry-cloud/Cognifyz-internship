```python
df.dropna(inplace=True)
# less than 1% of data was missing so deleting was better option
```

```python
df['Rating color'].value_counts()
```

```python
df['Rating text'].value_counts()
```

```python
df['Rating text'] = df['Rating text'].map({'Not rated': 0,
                                           'Poor': -1,
                                           'Average': 1,
                                           'Good': 2,
                                           'Very Good': 3,
                                           'Excellent': 4})
df.head() #mapping
```

```python
for col in ['Has Table booking','Has Online delivery','Is delivering now','Switch to order menu']:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

df.head()
```

```python
df.drop(columns=['Rating color'], inplace=True)
df.head() # both columns were representing the same data.
```

```python
ratings = df['Aggregate rating']
ratings.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

ratings.plot(kind='hist', bins=20, color='Purple', density=True)
plt.grid(axis='y')
plt.show() # bimodal
```
